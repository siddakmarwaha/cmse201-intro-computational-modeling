### Relation between height and dribbling

```python
plt.scatter(fifa22['dribbling'].head(200),fifa22['height_cm'].head(200))

plt.xlabel('Dribbling Ability (Higher is better)')
plt.ylabel('Height (in cm)')
plt.title('Relation between height and dribbling in Soccer')
plt.grid()
plt.show()
#Shorter player= Better dribbling
```

```python
x=fifa22['dribbling'].head(200)
y=fifa22['height_cm'].head(200)
```

```python
x.dropna(inplace=True) #Dropping NaN values
y=[]
for i in x.index:
    y.append(fifa22['height_cm'][i]) #Appending only those indices in y where x is not NaN
len(y)
```

```python
model = np.poly1d(np.polyfit(list(x), list(y), 2))

#added fitted polynomial line to scatterplot
polyline = np.linspace(58, 95, 50)
plt.scatter(x, y)
plt.plot(polyline, model(polyline),'r')
plt.xlabel('Dribbling Ability (Higher is better)')
plt.ylabel('Height (in cm)')
plt.grid()
plt.show()
```

### Relation between weight and pace

```python
x,y=fifa22['pace'].head(500),fifa22['weight_kg'].head(500)
x.dropna(inplace=True)
y=[]
for i in x.index:
    y.append(fifa22['weight_kg'][i]) #Appending only those indices in y where x is not NaN
len(y)
```

```python
plt.scatter(fifa22['pace'].head(500),fifa22['weight_kg'].head(500))
model = np.poly1d(np.polyfit(list(x), list(y), 2))

#add fitted polynomial line to scatterplot
polyline = np.linspace(30, 100, 50)
plt.scatter(x, y,color='b')
plt.plot(polyline, model(polyline),'red')
plt.xlabel('Pace Value')
plt.ylabel('Weight (in kg)')

plt.grid()
plt.show()
```

```python
premier_league=fifa22[fifa22['league_name']=='English Premier League']
```

```python
premier_league['age'].value_counts()
```

```python
premier_league.index=np.arange(0,652)
```

```python
shooting_of_pl_st=[]
age_of_pl_st=[]
overall_of_pl_st=[]
for i in range(652):
    if premier_league['player_positions'][i][0:2]=='ST':
        age_of_pl_st.append(premier_league['age'][i]) #Making list of PL players' age
        shooting_of_pl_st.append(premier_league['shooting'][i]) #Making list of PL players' shooting ability
        overall_of_pl_st.append(premier_league['overall'][i]) #Making list of PL players' overall rating
    else:
        pass
```

### Relation between age and shooting

```python
x,y=age_of_pl_st,shooting_of_pl_st
```

```python
plt.scatter(x,y)
model = np.poly1d(np.polyfit(list(x), list(y), 3))

#add fitted polynomial line to scatterplot
polyline = np.linspace(17, 38, 50)
plt.plot(polyline, model(polyline),'red')
plt.xlabel('Age (in kg)')
plt.ylabel('Shooting Ability')

plt.grid()
plt.show()
```

## When is players' prime?

### Here, the prime of PL players seems to be at around 32

```python
plt.scatter(age_of_pl_st,overall_of_pl_st)
model = np.poly1d(np.polyfit(age_of_pl_st, overall_of_pl_st, 2))

#add fitted polynomial line to scatterplot
polyline = np.linspace(17, 37, 50)
plt.plot(polyline, model(polyline),'red')
plt.xlabel('Age (in kg)')
plt.ylabel('Overall')
plt.grid()
plt.show()
```

```python
polyline[model(polyline).argmax()] #At this age, the curve is the highest
```
