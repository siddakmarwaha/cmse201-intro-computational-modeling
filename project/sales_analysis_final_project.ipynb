# Sales Compared with population to determine where they're doing well
<font size="3"> 
(CA is taken as benchmark, for eg: If 74337 sales in 39.51m population is normal, then...:)

```python
#[4] Learnt this code from a website linked in the citations
data1 = population
data2 = sales['State'].value_counts()

fig, ax1 = plt.subplots()

color = 'green'
ax1.set_xlabel('State')
ax1.set_ylabel('Population (in Millions)', color=color)
ax1.plot(sales['State'].value_counts().index, data1, color=color)
ax1.tick_params(axis='y', labelcolor=color)

ax2 = ax1.twinx()  # instantiate a second axes that shares the same x-axis

color = 'blue'
ax2.set_ylabel('Sales by State', color=color)  # we already handled the x-label with ax1
ax2.plot(sales['State'].value_counts().index, data2, color=color)
ax2.tick_params(axis='y', labelcolor=color)
plt.title('Sales Compared With Population in Each State')
fig.tight_layout()  # otherwise the right y-label is slightly clipped
plt.show()
```

<font size="3"> 
This graph might be a little tougher to understand. Now, this graph has two y-axes, one is the state's population (in millions), the other is the sales in that state. I compared these two things because in the previous bar graph, we couldn't tell if the company was really doing better in California than in New York because New York has way lesser population than California. So in this graph, we see the two lines meet at the start. This is because we have set the sales/population ratio of California to be the benchmark. If in any state the sales/population ratio is higher, the blue line will be above the green line (For eg: New York, Massachusets, Oregon). However, if in any state the sales/population is lower, the blue line will be below the green line (For eg: Texas, Georgia).
