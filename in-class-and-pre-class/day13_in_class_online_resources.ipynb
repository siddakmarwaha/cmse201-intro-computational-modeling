```python
#Write the code for your visualizations here
#Source: I tinkered and found this on my Day-13 PreClass Project and felt like it was
import matplotlib.pyplot as plt
import pandas as pd
ipl=pd.read_csv('matches.csv',)
ipl=ipl.drop(columns=['date', 'venue','umpire1','umpire2','umpire3'])
plt.figure(figsize=(8,8))
plt.plot(ipl['winner'].value_counts())
plt.plot(ipl['toss_winner'].value_counts(),'.')
plt.legend(['No. of Games Won','No. of Times Toss Won'])
plt.xlabel('Teams')
plt.ylabel('Frequency')
plt.xticks(rotation=90)
t=("The graph shows that there seems\n to be a corelation between " 
   "\nwho wins the toss before\n the cricket match and who wins the"
   "\n match.")
plt.text(14.5,80,t,ha='right',wrap=True,fontsize=15,style='italic',family='Times New Roman') #this is the tool
plt.show()
#Every member of your group should have this code in their final notebooks.
```

```python
batting_second=ipl[ipl['win_by_wickets']!=0]
```

```python
plt.figure(figsize=(7,7))
plt.hist(batting_second['win_by_wickets'],bins=30)  #Plotting the histogram of number of wickets the matches were won by.
plt.xlabel('Number of wickets match won by')
plt.ylabel('Frequency')
plt.grid(True)
plt.show() #We see a normal curve.
```
